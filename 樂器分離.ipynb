```python
#成功
import os
import subprocess
import shutil


def final_test_separation():
    # 明確指定 FFmpeg 路徑
    ffmpeg_path = r"C:\Users\jt725\AppData\Local\Microsoft\WinGet\Packages\Gyan.FFmpeg_Microsoft.Winget.Source_8wekyb3d8bbwe\ffmpeg-8.0.1-full_build\bin"
    os.environ['PATH'] = ffmpeg_path + os.pathsep + os.environ['PATH']
    
    # 設定 FFmpeg 環境變數 (某些程式需要這個)
    os.environ['FFMPEG_BINARY'] = os.path.join(ffmpeg_path, 'ffmpeg.exe')
    
    # --- 1. 設定路徑 ---
    input_wav = r"D:\音樂譜\ヨルシカ - 修羅（OFFICIAL VIDEO） (online-audio-converter.com).wav"
    output_folder = r"C:\Users\jt725\test_ai_output"

    if not os.path.exists(input_wav):
        print(f"錯誤:找不到檔案 {input_wav}")
        return

    os.makedirs(output_folder, exist_ok=True)

    # --- 2. 檔案更名 ---
    temp_name = "temp_test.wav"
    
    try:
        shutil.copy(input_wav, temp_name)
        print(f"已複製檔案到: {temp_name}")
    except Exception as e:
        print(f"複製檔案失敗: {e}")
        return
    
    print(f"正在處理:{temp_name}")
    
    # --- 3. 測試 FFmpeg ---
    print("\n[測試] 檢查 FFmpeg...")
    try:
        test = subprocess.run(['ffmpeg', '-version'], 
                            capture_output=True, text=True, timeout=5)
        if test.returncode == 0:
            print("[OK] FFmpeg 可用")
            # 顯示第一行版本資訊
            print(test.stdout.split('\n')[0])
        else:
            print(f"[警告] FFmpeg 測試失敗: {test.returncode}")
    except Exception as e:
        print(f"[錯誤] FFmpeg 測試異常: {e}")
        return
    
    # --- 4. 執行分離 ---
    cmd = ['demucs', '-d', 'cuda', '-n', 'mdx_extra', temp_name, '-o', output_folder]
    
    print(f"\n執行 demucs...")
    print("=" * 50)
    
    try:
        # 即時顯示輸出
        process = subprocess.Popen(
            cmd,
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            text=True,
            bufsize=1,
            universal_newlines=True
        )
        
        # 逐行顯示進度
        for line in process.stdout:
            print(line, end='')
        
        process.wait()
        
        if process.returncode == 0:
            print("\n" + "=" * 50)
            print("[成功] 分離完成!")
            
            expected_path = os.path.join(output_folder, "mdx_extra", "temp_test")
            if os.path.exists(expected_path):
                print(f"\n輸出位置: {expected_path}")
                print("\n音軌檔案:")
                for f in os.listdir(expected_path):
                    filepath = os.path.join(expected_path, f)
                    size = os.path.getsize(filepath) / 1024 / 1024
                    print(f"  ✓ {f} ({size:.2f} MB)")
            else:
                print("警告: 找不到輸出檔案")
        else:
            print(f"\n[失敗] 錯誤代碼: {process.returncode}")
            
    except Exception as e:
        print(f"執行異常: {e}")
    finally:
        if os.path.exists(temp_name):
            os.remove(temp_name)
            print(f"\n已清除暫存檔")

if __name__ == "__main__":
    final_test_separation()
```
